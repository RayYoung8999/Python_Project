```python
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda skill_list: ast.literal_eval(skill_list) if pd.notna(skill_list) else skill_list)
```

```python
df.head(3)
```

```python
df['job_work_from_home'].value_counts().plot(kind='pie', startangle=90, autopct='1.1f')
plt.title('Work from Home Status')
plt.show()
```

```python
df['job_title_short'].value_counts().plot(kind='pie')
plt.title('Job Porpotions')
plt.show()
```

```python
# create fig, ax >> define dict (for index of df and each chart's title) >> for loop and plot:
fig, ax = plt.subplots(1,3)

job_dict = {
'job_work_from_home' : 'Work Status',
'job_no_degree_mention' : 'Degree Req',
'job_health_insurance' : 'Health insurance'
}

for i, (column, title) in enumerate(job_dict.items()):
    ax[i].pie(df[column].value_counts(), startangle=90, autopct='%1.1f%%', labels=[False, True])
    ax[i].set_title(title)

plt.show()



 
```
